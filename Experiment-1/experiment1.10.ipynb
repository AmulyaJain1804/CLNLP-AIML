# Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("employee_information_100.csv")
```

```python
plt.figure(figsize=(8, 5))
plt.hist(data['Age'], bins=10, color='lightgreen', edgecolor='black')
plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
