# Experiment 1.4: Draw a histogram showing salary distribution.

```python
import pandas as pd
```

```python
data = pd.read_csv("employee_information_100.csv")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))
plt.hist(data['Salary'], bins=10, color='skyblue', edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```
