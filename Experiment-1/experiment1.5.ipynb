# Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
import pandas as pd
```

```python
data = pd.read_csv("employee_information_100.csv")
```

```python
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(data['Experience_Years'], data['Salary'], alpha=0.7)
plt.title('Experience vs Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.show()
```
