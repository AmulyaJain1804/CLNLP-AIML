# Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("employee_information_100.csv")
```

```python
overall_avg_salary = data['Salary'].mean()
above_avg = data[data['Salary'] > overall_avg_salary][['Name', 'Department', 'Salary']]
print(above_avg)
```
