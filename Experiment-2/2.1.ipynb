Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
Convert text to lowercase. (Also display how many uppercase?)
Remove punctuation.        (Also display how many different types of punctuations?)
Remove numbers.            (Also display how many numbers?)
Remove extra whitespace. (How much extra whitespace is present?)
Display the cleaned text.   

```python
with open("2.1_text_data.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

```python
countLow = 0
countHigh = 0
for char in content:
    if char.islower():
        countLow += 1
    else:
        countHigh += 1
print("The no. of lowercase characters: ", countLow)
print("The no. of highcase characters: ", countHigh)
```

```python
print("Original Text: -")
print(content)
print("-" * 200)
content_lower = content.lower()
print("All Lowercase characters: -")
print(content_lower)
```

```python
import string
unique_punctuations = set()
for char in content_lower:
    if char in string.punctuation:
        unique_punctuations.add(char)

print("No. of unique punctuations: ", len(unique_punctuations))
print(unique_punctuations)
```

```python
content_no_punctuations = ""
for char in content_lower:
    if char not in string.punctuation:
        content_no_punctuations += char
print("Text without punctuations: -")
print(content_no_punctuations)
```

```python
count_number = 0
for char in content_no_punctuations:
    if char.isdigit():
        count_number += 1
print("No. of numeric digits in content: ", count_number)
```

```python
content_no_numbers = ""
for char in content_no_punctuations:
    if not char.isdigit():
        content_no_numbers += char
print("Text without nunbers and punctuations: -")
print(content_no_numbers)
```

```python
total_spaces = 0
for char in content_no_numbers:
    if char.isspace():
        total_spaces += 1
words_list = content_no_numbers.split()

print("Total Spaces: ", total_spaces)
print("Words list present in content: -")
print(words_list)
```

```python
final_clean_text = " ".join(words_list)
```

```python
ideal_spaces = max(0, len(words_list) - 1)
extra_whitespace = total_spaces - ideal_spaces

print("Extra whitespace characters present: ", extra_whitespace)
print("-" * 50)
print("Final Cleaned Text: -")
print(final_clean_text)
```
