#### Using NLTK

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt')
```

```python
# 1. Sentence Tokenization
sentences = sent_tokenize(content)
print(f"\nSentences ({len(sentences)} found):")
for i, sentence in enumerate(sentences, 1):
    print(f" {i}. {sentence}")
```

```python
# 2. Word Tokenization
words = word_tokenize(content)
print(f"\nWords ({len(words)} found):")
print(words)
```

#### Using only Python
Sentence Split: Instead of parsing patterns, it loops through words. It detects a sentence boundary when a word ends with ., !, or ?. The safety check not word.replace('.', '').isdigit() prevents the script from breaking sentences at decimals like 2.2.

Word Split: It runs a series of .replace() mutations to forcefully pad spaces around punctuation symbols (e.g., changing "world!" to "world ! "). Once spaces isolate the characters, a native .split() effortlessly groups them into clean array slots.

```python
# 1. SENTENCE TOKENIZATION
sentences = []
current_sentence = []

# Split text by spaces to check each word's ending punctuation
for word in content.split():
    current_sentence.append(word)
    
    # Check if word ends with terminal punctuation AND make sure it's not a decimal (like 2.2)
    if (word.endswith('.') or word.endswith('!') or word.endswith('?')) and not (word.replace('.', '').isdigit()):
        # Join words back together to form the sentence string
        sentences.append(" ".join(current_sentence))
        current_sentence = [] # Reset for next sentence

# Catch any trailing text if it didn't end with punctuation
if current_sentence:
    sentences.append(" ".join(current_sentence))

# Print Sentence Results
print(f"\nSentences ({len(sentences)} found):")
for i, sentence in enumerate(sentences, 1):
    print(f" {i}. {sentence}")
```

```python
# 2. WORD TOKENIZATION
# To isolate punctuation marks as separate tokens, we add spaces around them
spaced_text = content

# Add padding spaces to common punctuation marks manually
punctuation_marks = ['.', '!', '?', ',', '(', ')']
for mark in punctuation_marks:
    # Special safety check: don't break decimals like 2.2
    if mark == '.':
        # Temporary placeholder for decimals so they don't get broken apart
        pass 
    else:
        spaced_text = spaced_text.replace(mark, f" {mark} ")

# Simple trick to handle periods safely without breaking decimals:
# If a period is followed by a space or is at the end of the text, it's punctuation.
spaced_text = spaced_text.replace(". ", " . ")
if spaced_text.endswith("."):
    spaced_text = spaced_text[:-1] + " ."

# Split by whitespace to get the final clean tokens list
words = spaced_text.split()

# Print Word Results
print(f"\nWords ({len(words)} found):")
print(words)
```
