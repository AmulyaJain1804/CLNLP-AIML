Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
with open("2.3_clean_data.txt", "r", encoding="utf-8") as file:
    content = file.read()
    print(content)
```

```python
stop_words = {"the", "is", "a", "an", "in", "of", "and", "to", "for", "on", "with", "from", "by", "this", "that", "it"}
```

```python
cleaned_text = ""
for char in content:
    if char.isalnum() or char.isspace():
        cleaned_text += char
    else:
        # Replace punctuation marks with spaces so words do not accidentally merge
        cleaned_text += " "
```

```python
original_tokens = cleaned_text.split()
```

```python
filtered_tokens = []
found_stop_words = set()

for token in original_tokens:
    if token.lower() in stop_words:
        found_stop_words.add(token.lower())
    else:
        filtered_tokens.append(token)
```

```python
print(f"\n1. Original Tokens ({len(original_tokens)} total):\n", original_tokens)
print(f"\n2. Filtered Tokens ({len(filtered_tokens)} total):\n", filtered_tokens)
print(f"\n3. Stop Words Identified Separately ({len(found_stop_words)} unique):\n", sorted(list(found_stop_words)))
```
