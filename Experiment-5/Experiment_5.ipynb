**(a) SentencePiece Tokenization using Pretrained Model**

```python
!pip install sentencepiece

import sentencepiece as spm

with open("/content/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()


!wget -q https://huggingface.co/google/mt5-small/resolve/main/spiece.model -O spiece.model

sp = spm.SentencePieceProcessor(model_file="spiece.model")

tokens = sp.encode(text, out_type=str)
token_ids = sp.encode(text, out_type=int)

print("========== (a) PRETRAINED SENTENCEPIECE ==========")
print("Tokens and Token IDs:")
print("--------------------------------")

for token, token_id in zip(tokens, token_ids):
    print(token, ":", token_id)
```

**5.3**

**Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.**
                            
                   **"The young student is reading an interesting book in the library."**   



### **(a) Using Spacy**

```python
!pip install spacy
!python -m spacy download en_core_web_sm

import spacy

nlp = spacy.load("en_core_web_sm")

text = "The young student is reading an interesting book in the library."

doc = nlp(text)

print("Token\t\tPOS\tDescription")
print("--------------------------------------------")

seen = set()

for token in doc:
    if token.text not in seen:
        print(f"{token.text}\t\t{token.pos_}\t{spacy.explain(token.pos_)}")
        seen.add(token.text)
```

### **(b) Using NLTK**

```python
!pip install nltk

import nltk

nltk.download("punkt")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("punkt_tab")
nltk.download("tagsets_json")

text = "The young student is reading an interesting book in the library."

tokens = nltk.word_tokenize(text)

pos_tags = nltk.pos_tag(tokens)

print("Token\t\tPOS\tDescription")
print("--------------------------------------------")

seen = set()

for token, tag in pos_tags:
    if token not in seen:
        print(f"{token}\t\t{tag}\t{nltk.help.upenn_tagset(tag)}")
        seen.add(token)
```

**5.4**
  
**Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.**


```python
!pip install spacy
!python -m spacy download en_core_web_sm

import spacy
from collections import Counter

nlp = spacy.load("en_core_web_sm")

with open("/content/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

doc = nlp(text)

frequency = Counter(
    token.text.lower()
    for token in doc
    if not token.is_punct and not token.is_space
)

print("Token\t\tPOS\tDescription\t\tFrequency")
print("------------------------------------------------------------")

seen = set()

for token in doc:
    word = token.text.lower()

    if not token.is_punct and not token.is_space and word not in seen:
        print(
            f"{word}\t\t"
            f"{token.pos_}\t"
            f"{spacy.explain(token.pos_)}\t\t"
            f"{frequency[word]}"
        )
        seen.add(word)
```

**5.5**

**Write a Python program to train a Word2Vec CBOW model using the given text corpus. Display the vector representation of a selected word and find the three most similar words.**


```python
!pip install gensim
from gensim.models import Word2Vec

with open("/content/sample_data/corpus.txt", "r") as file:
    sentences = [line.lower().split() for line in file]

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0
)

word = "learning"

print("Vector representation of", word, ":")
print(model.wv[word])

print("\nThree most similar words:")
similar_words = model.wv.most_similar(word, topn=3)

for w, similarity in similar_words:
    print(w, ":", similarity)
```

**Train a Word2Vec model using the provided corpus and calculate the cosine similarity between selected word pairs.**

```python
from gensim.models import Word2Vec

with open("/content/sample_data/corpus.txt", "r") as file:
    sentences = [line.lower().split() for line in file]

model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=0
)

pairs = [
    ("machine", "learning"),
    ("natural", "language"),
    ("deep", "learning")
]

for word1, word2 in pairs:
    similarity = model.wv.similarity(word1, word2)
    print(f"Similarity between '{word1}' and '{word2}': {similarity:.4f}")
```
